```python
# Define the Queue class
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        return None

    def is_empty(self):
        return len(self.items) == 0

# Create a Queue object and test it
q = Queue()
q.enqueue("apple")
q.enqueue("banana")
q.enqueue("cherry")

print("Dequeued:", q.dequeue())     # Output: apple
print("Dequeued:", q.dequeue())     # Output: banana
print("Is empty?", q.is_empty())    # Output: False
print("Dequeued:", q.dequeue())     # Output: cherry
print("Is empty?", q.is_empty())    # Output: True
```
