# Lab 10 Task 1 — Refactor Nested Conditionals

This notebook refactors a nested conditional `discount` function into a
cleaner, more readable implementation using helper functions and input
validation. It includes examples and simple assertions to demonstrate
correct behavior.

```python
from typing import Union

Number = Union[int, float]

def _apply_discount(price: float, rate: float) -> float:
    """Apply a percentage discount rate to a price and return the result."""
    return price * (1 - rate)

def discount(price: Number, category: str) -> float:
    """Return the discounted price based on category and thresholds.

    Rules:
    - If ``category`` is "student":
      - price > 1000 -> 10% off
      - otherwise -> 5% off
    - For all other categories:
      - price > 2000 -> 15% off
      - otherwise -> no discount

    The function validates inputs and raises appropriate errors for
    incorrect types or negative prices.

    Parameters
    ----------
    price : int or float
        Original price (must be non-negative).
    category : str
        Customer category string (case-insensitive).

    Returns
    -------
    float
        Discounted price as a float.
    """
    # Input validation
    if not isinstance(price, (int, float)) or isinstance(price, bool):
        raise TypeError(f"price must be int or float, got {type(price).__name__}")
    if price < 0:
        raise ValueError("price must be non-negative")
    if not isinstance(category, str):
        raise TypeError(f"category must be str, got {type(category).__name__}")

    cat = category.strip().lower()

    if cat == "student":
        rate = 0.10 if price > 1000 else 0.05
    else:
        rate = 0.15 if price > 2000 else 0.0

    return float(_apply_discount(float(price), rate))

# Simple demonstration and tests
if __name__ == "__main__":
    examples = [
        (1200, 'student', 1200 * 0.9),    # student above 1000 -> 10% off
        (800, 'student', 800 * 0.95),     # student below 1000 -> 5% off
        (2500, 'regular', 2500 * 0.85),   # non-student above 2000 -> 15% off
        (1500, 'regular', 1500.0),        # non-student below 2000 -> no discount
    ]

    for price, cat, expected in examples:
        result = discount(price, cat)
        print(f"discount({price}, '{cat}') = {result} (expected {expected})")
        assert abs(result - expected) < 1e-9, f"Mismatch for {price}, {cat}"

    # Error cases
    try:
        discount(-10, 'student')
    except ValueError as e:
        print('Caught expected ValueError for negative price')
    else:
        raise AssertionError('Negative price did not raise ValueError')

    try:
        discount(100, 123)  # invalid category type
    except TypeError as e:
        print('Caught expected TypeError for non-string category')
    else:
        raise AssertionError('Non-string category did not raise TypeError')

    print('All checks passed.')
```

# Lab 10 Task 2 — Optimize Redundant Loops

This section refactors a naive nested-loop implementation of
`find_common(a, b)` into an optimized version using Python sets
to compute the intersection. The set-based approach is cleaner and
more efficient for large inputs (average O(n) time vs O(n*m) for
nested loops). A deterministic sorted list is returned when possible.

```python
from typing import Iterable, List, TypeVar

T = TypeVar('T')

def find_common(a: Iterable[T], b: Iterable[T]) -> List[T]:
    """Return the common elements of `a` and `b` as a list."""
    # Use set intersection for efficiency
    set_a = set(a)
    set_b = set(b)
    common = set_a & set_b
    try:
        # Return sorted list for deterministic output when elements are orderable
        return sorted(common)
    except TypeError:
        # If elements are not orderable (e.g., dicts), return list without sorting
        return list(common)

# Demonstration and simple checks
if __name__ == '__main__':
    a = [1, 2, 3, 4, 5]
    b = [4, 5, 6, 7]
    print('Common elements:', find_common(a, b))  # -> [4, 5]

    # Edge cases
    assert find_common([], [1,2,3]) == []
    assert find_common([1,1,2], [1]) == [1]
    assert set(find_common([None, True], [True, False])) == {True}
    print('find_common checks passed.')
```
