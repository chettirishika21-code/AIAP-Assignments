```python
def divide_numbers(a, b):
    """
    Divides two numbers and handles division by zero.

    Parameters:
    a (float): Numerator
    b (float): Denominator

    Returns:
    float or str: Result of division if successful, or an error message if division by zero occurs.
    """
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."

# Sample execution
result = divide_numbers(10, 0)
print(result)
```
