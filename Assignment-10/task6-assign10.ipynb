```python
def grade(score):
    """
    Returns a letter grade based on the numeric score.
    """
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

# Sample execution
test_score = 85
result = grade(test_score)
print(f"Score: {test_score}, Grade: {result}")
```
