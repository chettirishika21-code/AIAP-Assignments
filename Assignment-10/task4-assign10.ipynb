```python
def calculate_average(scores):
    """
    Calculates the average of the given scores.
    """
    total = sum(scores)
    return total / len(scores)

def find_highest(scores):
    """
    Finds the highest score in the list.
    """
    highest = scores[0]
    for s in scores:
        if s > highest:
            highest = s
    return highest

def find_lowest(scores):
    """
    Finds the lowest score in the list.
    """
    lowest = scores[0]
    for s in scores:
        if s < lowest:
            lowest = s
    return lowest

def process_scores(scores):
    """
    Processes the scores by calculating and displaying average, highest, and lowest values.
    """
    avg = calculate_average(scores)
    highest = find_highest(scores)
    lowest = find_lowest(scores)

    print(f"Average: {avg:.2f}")
    print(f"Highest: {highest}")
    print(f"Lowest: {lowest}")

# 👇 Sample execution
scores = [85, 92, 78, 90, 88]
process_scores(scores)
```
