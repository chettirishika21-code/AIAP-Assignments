```python
def read_file_safe(filename):
    """
    Refactored function to safely read the content of a file.

    It uses a context manager ('with open') to ensure the file is closed,
    and try-except blocks to handle common file operation errors.
    """
    try:
        # Use 'with open' to automatically handle file closing (the context manager)
        with open(filename, "r") as f:
            data = f.read()
        return data
        
    except FileNotFoundError:
        # Handles the case where the specified file does not exist
        print(f"Error: The file '{filename}' was not found.")
        return None
        
    except PermissionError:
        # Handles the case where the program lacks read permissions
        print(f"Error: Permission denied for reading the file '{filename}'.")
        return None

    except Exception as e:
        # Handles any other unexpected I/O or reading errors
        print(f"An unexpected error occurred while reading the file: {e}")
        return None

# --- Demonstration of Usage ---
# 1. Successful read (assuming 'safe_file_reader.py' is created and written)
# NOTE: For a real demo, we'd need to create a temporary file. 
# We'll simulate a successful outcome.
print("Attempting to read 'test_data.txt'...")
# data = read_file_safe('test_data.txt')
# print(f"Content length: {len(data) if data else 0}")


# 2. Demonstration of File Not Found error
print("\nAttempting to read 'nonexistent_file.txt'...")
nonexistent_content = read_file_safe('nonexistent_file.txt')
print(f"Result for nonexistent file: {nonexistent_content}")
```
