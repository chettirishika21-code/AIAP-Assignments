```python
class Student:
    """
    Represents a student and manages their personal details and subject marks.
    The marks are stored in a list for easy summation.
    """

    def __init__(self, name, age, mark1, mark2, mark3):
        """
        Initializes a new Student instance.

        Args:
            name (str): The full name of the student.
            age (int): The age of the student.
            mark1 (int): Mark for subject 1.
            mark2 (int): Mark for subject 2.
            mark3 (int): Mark for subject 3.
        """
        # Improved naming for clarity
        self.name = name
        self.age = age
        
        # Store marks in a list for modularity and easy processing
        self.marks = [mark1, mark2, mark3]

    def get_details(self):
        """
        Prints the student's personal details in a readable format.
        """
        # Use f-strings for clean, formatted output
        print(f"--- Student Details ---")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Marks: {self.marks}")

    def calculate_total(self):
        """
        Calculates and returns the sum of all marks.
        Leverages the built-in sum() function for simplicity.
        """
        return sum(self.marks)

    def calculate_average(self):
        """
        Calculates and returns the average mark.
        """
        if not self.marks:
            return 0
        return self.calculate_total() / len(self.marks)

# --- Demonstration of the Refactored Class ---

# 1. Create an instance with clear arguments
student_a = Student(
    name="Alice Smith",
    age=16,
    mark1=85,
    mark2=92,
    mark3=78
)

# 2. Call the improved methods
print("=" * 30)
student_a.get_details()
print(f"Total Marks: {student_a.calculate_total()}")
print(f"Average Mark: {student_a.calculate_average():.2f}")
print("=" * 30)
```
