```python
# --- Refactored Solution 1: Dictionary Dispatch (Best for performance/scalability) ---
# This approach uses a central dictionary to map shape names (keys) to handler functions (values).

import math

# 1. Define specific, small, and clear calculation functions
def _area_rectangle(length, width):
    """Calculates the area of a rectangle (requires length and width)."""
    return length * width

def _area_square(side, _): # The underscore (_) indicates an unused second parameter
    """Calculates the area of a square (only needs the side length)."""
    return side * side

def _area_circle(radius, _): # The second parameter is unused
    """Calculates the area of a circle (uses radius as the first parameter)."""
    return math.pi * radius * radius

# 2. Create the dispatch dictionary
# Maps the shape name (string) to the corresponding function
AREA_CALCULATORS = {
    "rectangle": _area_rectangle,
    "square": _area_square,
    "circle": _area_circle,
}

# 3. Create the simplified main dispatcher function
def calculate_area_dispatch(shape, x, y=0):
    """
    Calculates the area of a shape using dictionary dispatch.
    x is typically the first dimension (side, radius, length).
    y is the optional second dimension (width).
    """
    # Retrieve the correct function handler or return an error message
    handler = AREA_CALCULATORS.get(shape.lower())
    
    if handler:
        # Call the appropriate function with the provided dimensions
        # Note: Circle and square functions will ignore the 'y' argument (using _)
        return handler(x, y)
    else:
        return f"Error: Unknown shape '{shape}'. Supported shapes are: {', '.join(AREA_CALCULATORS.keys())}"


# --- Demonstration ---
print("--- Solution 1: Dictionary Dispatch Output ---")
print(f"Rectangle (4, 5): {calculate_area_dispatch('rectangle', 4, 5)}")
print(f"Square (6): {calculate_area_dispatch('square', 6)}")
print(f"Circle (3): {calculate_area_dispatch('circle', 3):.2f}")
print(f"Unknown shape: {calculate_area_dispatch('triangle', 3, 5)}\n")


# --- Refactored Solution 2: Modular Class Design ---
# This object-oriented approach is cleaner for complex applications
# and avoids the single large dispatch function.

class ShapeCalculator:
    
    @staticmethod
    def rectangle(length, width):
        """Calculates area of a rectangle."""
        return length * width

    @staticmethod
    def square(side):
        """Calculates area of a square."""
        return side * side

    @staticmethod
    def circle(radius):
        """Calculates area of a circle."""
        return math.pi * radius * radius

# --- Demonstration ---
print("--- Solution 2: Modular Class Output ---")
print(f"Rectangle (4, 5): {ShapeCalculator.rectangle(4, 5)}")
print(f"Square (6): {ShapeCalculator.square(6)}")
print(f"Circle (3): {ShapeCalculator.circle(3):.2f}")

# The user-facing code is much cleaner:
# area = ShapeCalculator.square(8)
```
