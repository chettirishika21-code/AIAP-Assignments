```python
class Stack:
    """
    A simple Last-In, First-Out (LIFO) stack implementation 
    using a Python list.
    """
    def __init__(self):
        """Initializes an empty stack."""
        self._items = []

    def push(self, item):
        """Adds an item to the top of the stack."""
        self._items.append(item)

    def pop(self):
        """
        Removes and returns the item at the top of the stack.
        
        Raises:
            IndexError: If the stack is empty.
        """
        if not self._items:
            raise IndexError("pop from empty stack")
        return self._items.pop()

    def peek(self):
        """
        Returns the item at the top of the stack without removing it.
        
        Raises:
            IndexError: If the stack is empty.
        """
        if not self._items:
            raise IndexError("peek from empty stack")
        return self._items[-1]

    def is_empty(self):
        """Returns True if the stack is empty, False otherwise."""
        return len(self._items) == 0

    def size(self):
        """Returns the number of items in the stack."""
        return len(self._items)

    def __len__(self):
        """Enables len(stack)"""
        return self.size()


# --- Code to run the stack and produce output ---

print("--- Testing Stack Operations ---")

# 1. Initialization and Push
my_stack = Stack()
my_stack.push("A")
my_stack.push("B")
my_stack.push("C")
print(f"1. Stack initialized and pushed 'A', 'B', 'C'.")
print(f"   Current size: {len(my_stack)}")
print(f"   Is empty? {my_stack.is_empty()}")

# 2. Peek (Check the top without removal)
top_item = my_stack.peek()
print(f"\n2. Peeking at the top: {top_item}")

# 3. Pop (Remove and return)
popped1 = my_stack.pop()
print(f"\n3. Popped first item: {popped1}")
print(f"   New size: {my_stack.size()}")

# 4. Final Pops to empty the stack
popped2 = my_stack.pop()
popped3 = my_stack.pop()
print(f"4. Popped remaining items: {popped2}, {popped3}")
print(f"   Final size: {len(my_stack)}")
print(f"   Is empty? {my_stack.is_empty()}")

# 5. Testing the IndexError edge case
print("\n5. Testing pop() on an empty stack...")
try:
    my_stack.pop()
except IndexError as e:
    print(f"   Caught expected error: {e}")
```
