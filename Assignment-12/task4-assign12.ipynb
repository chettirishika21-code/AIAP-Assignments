```python
def f(x):
    """The function to minimize: f(x) = 2*x^3 + 4*x + 5"""
    return 2 * (x**3) + 4 * x + 5

def gradient(x):
    """The derivative: f'(x) = 6*x^2 + 4"""
    return 6 * (x**2) + 4

def gradient_descent(start_x, lr, iters, max_x_limit):
    x = start_x
    print(f"--- Starting GD with x_start = {x} ---")
    
    for i in range(iters):
        # 1. Check for overflow condition before calculation (stops silently)
        if abs(x) > max_x_limit:
            break 
            
        grad = gradient(x)
        x -= lr * grad
        
    return x

# --- Parameters and Execution ---
X_START = 1.0        
LEARNING_RATE = 0.01 
ITERATIONS = 100     
MAX_ABS_X = 100.0  # Safety threshold to prevent x^3 from overflowing

final_x = gradient_descent(X_START, LEARNING_RATE, ITERATIONS, MAX_ABS_X)

print("\n------------------------------------")
print("Optimization Complete.")

try:
    final_f = f(final_x)
    print(f"Final x value: {final_x:.5f}")
    print(f"Final f(x) value: {final_f:.5f}")
    print("\nNote: The function f(x) has no local minimum, so x moves towards -infinity.")
except OverflowError:
    # This handles the unlikely case where the final calculated x still causes an overflow
    print("Final x value: Exceeded safe limit.")
    print("Final f(x) value: Too large to calculate (overflow).")
    print("\nNote: The function f(x) has no local minimum, so x moves towards -infinity.")
```
