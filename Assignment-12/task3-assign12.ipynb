```python
# This script solves the Linear Programming problem using a brute-force
# search over all feasible integer combinations, avoiding external libraries like pulp.

# Problem Formulation:
# Maximize Profit (Z) = 6*A + 5*B
# Subject to:
# 1. Milk: A + B <= 5
# 2. Choco: 3*A + 2*B <= 12
# A >= 0, B >= 0, A and B are Integers.

def solve_chocolate_optimization():
    """
    Finds the optimal integer units of A and B to maximize profit.
    """
    max_profit = -1  # Initialize with a value lower than any possible profit
    optimal_A = 0
    optimal_B = 0

    # Determine the search space:
    # From Milk constraint (A + B <= 5), the max value for A or B is 5.
    # From Choco constraint (3A + 2B <= 12), the max value for A is 4 (if B=0).
    # The absolute maximum unit count for A or B must be less than 6.

    # Iterate through all possible non-negative integer values for A and B
    # that are within a reasonable bound (e.g., up to 5 or 6, based on constraints).
    for A in range(6):  # A can't exceed 5 (due to Milk) or 4 (due to Choco)
        for B in range(7): # B can't exceed 5 (due to Milk) or 6 (due to 2B <= 12)

            # Check if the combination (A, B) is feasible (meets all constraints)

            # Constraint 1: Milk (A + B <= 5)
            if A + B > 5:
                continue # Skip this combination

            # Constraint 2: Choco (3*A + 2*B <= 12)
            if 3 * A + 2 * B > 12:
                continue # Skip this combination

            # If both constraints are met, calculate the profit
            current_profit = 6 * A + 5 * B

            # Check if this profit is better than the maximum found so far
            if current_profit > max_profit:
                max_profit = current_profit
                optimal_A = A
                optimal_B = B

    # --- Print Results ---
    print("--- Chocolate Production Optimization Results (Algebraic Solution) ---")
    print(f"Status: Optimal (Found by exhaustive search of integer points)")
    print("\nOptimal Solution:")
    print(f"Units of Chocolate A to produce: {optimal_A} units")
    print(f"Units of Chocolate B to produce: {optimal_B} units")
    print(f"\nMaximum Profit: Rs {max_profit}")

    # Optional: Resource check
    print("\nResource Utilization:")
    print(f"Milk Used: {optimal_A + optimal_B} out of 5 available.")
    print(f"Choco Used: {3*optimal_A + 2*optimal_B} out of 12 available.")

# Execute the function
solve_chocolate_optimization()
```
