```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return "Stack is empty"

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return "Stack is empty"

    def is_empty(self):
        return len(self.items) == 0

# ✅ Sample Execution
stack = Stack()
stack.push(10)
stack.push(20)
print("Top item:", stack.peek())     # Output: Top item: 20
print("Popped item:", stack.pop())   # Output: Popped item: 20
print("Is empty?", stack.is_empty()) # Output: Is empty? False
```
